```python
from sqlalchemy import create_engine, text

engine = create_engine("mysql+pymysql://root:2828sms**@localhost:3306/olist")
```

```python
tables = {
    "orders": orders,
    "customers": costumers,
    "products": products,
    "sellers": sellers,
    "reviews": reviews,
    "translation": translation,
    "order_items": order_items,
    "payments": payments
}

for table_name, df in tables.items():
    df.to_sql(
        table_name,
        engine,
        if_exists="replace",
        index=False,
        chunksize=2000
    )
```
